# seed_customers

Writes a small SYNTHETIC `customers` Delta table into the default lakehouse. FabricOps sandbox only.
Parameter `id_column` is the name of the key column; the schema-drift injector re-runs this notebook with another name.

```python
id_column = "customer_id"
```

```python
countries = ["AU", "NZ", "SG", "GB", "US"]
segments = ["consumer", "small_business", "enterprise"]
rows = [
    (
        i,
        f"Customer {i:04d}",
        countries[i % len(countries)],
        segments[i % len(segments)],
        f"2025-{1 + i % 12:02d}-{1 + i % 28:02d}",
        float(100 + (i * 37) % 900),
    )
    for i in range(1, 51)
]
columns = [id_column, "full_name", "country", "segment", "signup_date", "lifetime_value"]
df = spark.createDataFrame(rows, columns)
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("customers")
print(f"seeded {df.count()} synthetic customers with key column {id_column!r}")
```
